```python
import pandas as pd
import numpy as np
import os
```

```python
idx = 0
col_names = [] #stores column names from the first text files
data_lines = [] #stores all data from 49 text files

directory = "Data_files" 

#To loop through all .txt files in the directory
for filename in os.listdir(directory):
    file_path = os.path.join(directory, filename)

    #To check if it is a file and ends with .txt
    if os.path.isfile(file_path) and filename.endswith(".txt"):
        with open(file_path, "r") as myfile:
            for idx, line in enumerate(myfile):
                line = line.strip()
                if not line:
                    continue
                
                split_line = line.split("\t") # breaks them up into strings
                if idx == 0:
                    col_names = split_line
                else:
                    data_lines.append(split_line)
                idx += 1 

#TO CONFIRM CORRECT DATA IS EXTRACTED
print(col_names)
print(data_lines)
```

```python
#Creating a DataFrame from the read data
df = pd.DataFrame(data=data_lines, columns=col_names)
```

```python
df
```

```python
wild_seq = "WildType.Sequence"
mutant_seq = "Mutant.Sequence"

# To find what mutation each data file has. For substituition we need to check all positions first.
def find_mutation(wild_seq, mutant_seq):
    if len(wild_seq) > len(mutant_seq):
        return "Deletion"
    elif len(wild_seq) < len(mutant_seq):
        return "Insertion"
    else:
        for i in range(len(wild_seq)):
            if wild_seq[i] != mutant_seq[i]:
                return "Substitution"
        else:
            return "No mutation" # returns this only after all conditions and positions are checked.

#Creating a new column for type of mutation in the data frame
df["Mutation_type"] = df.apply(lambda row: find_mutation(row ["WildType.Sequence"], row ["Mutant.Sequence"]), axis=1)

# Displaying the results
df
```

```python
# Check if the mutation is in the promoter region or the coding region
def mutation_location(wild_seq, mutant_seq):
    # Split the wild-type sequence into promoter and coding regions
    promoter_region = wild_seq[:1000]  # First 1000 nucleotides are the promoter region
    coding_region = wild_seq[1000:]    # Everything after the first 1000 nucleotides is the coding region
    
    if mutant_seq[:1000] != promoter_region:  # If the first 1000 nucleotides are different
        return 'Promoter Region'
    elif mutant_seq[1000:] != coding_region:  # If the part after 1000 nucleotides is different
        return 'Coding Region'
    else:
        return 'No mutation'  # If no change in both regions

# Apply the function to each row in the DataFrame
df["Mutation_location"] = df.apply(lambda row: mutation_location(row["WildType.Sequence"], row ["Mutant.Sequence"]), axis=1)

# Display the updated DataFrame
df
```

```python
# error message resulted in converting mean values from strings into floating numbers 
#Calculating cell viability mean of both mutant and wild type
df['WT_CellViability_Mean'] = df[['CellViability.WT.Rep1', 'CellViability.WT.Rep2', 'CellViability.WT.Rep3']].astype(float).mean(axis=1)
df['Mut_CellViability_Mean'] = df[['CellViability.Mut.Rep1', 'CellViability.Mut.Rep2', 'CellViability.Mut.Rep3']].astype(float).mean(axis=1)

#Calculate mean mRNA expression levels
df['WT_mRNA_Mean'] = df[['mRNA.Expression.WT.Rep1', 'mRNA.Expression.WT.Rep2', 'mRNA.Expression.WT.Rep3']].astype(float).mean(axis=1)
df['Mut_mRNA_Mean'] = df[['mRNA.Expression.Mut.Rep1', 'mRNA.Expression.Mut.Rep2', 'mRNA.Expression.Mut.Rep3']].astype(float).mean(axis=1)

# Calculate mean protein expression levels
df['WT_Protein_Mean'] = df[['Protein.Expression.WT.Rep1', 'Protein.Expression.WT.Rep2', 'Protein.Expression.WT.Rep3']].astype(float).mean(axis=1)
df['Mut_Protein_Mean'] = df[['Protein.Expression.Mut.Rep1', 'Protein.Expression.Mut.Rep2', 'Protein.Expression.Mut.Rep3']].astype(float).mean(axis=1)

# Calculate percentage changes
df['mRNA_%Change'] = ((df['Mut_mRNA_Mean'] - df['WT_mRNA_Mean']) / df['WT_mRNA_Mean']) * 100
df['Protein_%Change'] = ((df['Mut_Protein_Mean'] - df['WT_Protein_Mean']) / df['WT_Protein_Mean']) * 100
df['CellViability_%Change'] = ((df['Mut_CellViability_Mean'] - df['WT_CellViability_Mean']) / df['WT_CellViability_Mean']) * 100

# Display the updated DataFrame
df
```

```python
# Fold changes (often used in gene expression analysis)
df['CellViability_FoldChange'] = df['Mut_CellViability_Mean'] / df['WT_CellViability_Mean'].replace(0, np.nan)
df['mRNA_FoldChange'] = df['Mut_mRNA_Mean'] / df['WT_mRNA_Mean'].replace(0, np.nan)
df['Protein_FoldChange'] = df['Mut_Protein_Mean'] / df['WT_Protein_Mean'].replace(0, np.nan)

df['CellViability_log2FC'] = np.log2(df['CellViability_FoldChange']).replace(0, np.nan)
df['mRNA_log2FC'] = np.log2(df['mRNA_FoldChange']).replace(0, np.nan)
df['Protein_log2FC'] = np.log2(df['Protein_FoldChange']).replace(0, np.nan)

df
```

```python
from scipy import stats

#This code was generated by AI
# Function to calculate p-value from t-test
def calculate_pvalue(row, metric):
    wt_values = [float(row[f'{metric}.WT.Rep1']), float(row[f'{metric}.WT.Rep2']), float(row[f'{metric}.WT.Rep3'])]
    mut_values = [float(row[f'{metric}.Mut.Rep1']), float(row[f'{metric}.Mut.Rep2']), float(row[f'{metric}.Mut.Rep3'])]
    _, p_value = stats.ttest_ind(wt_values, mut_values)
    return p_value

# Calculate p-values for each metric
df['CellViability_pvalue'] = df.apply(lambda row: calculate_pvalue(row, 'CellViability'), axis=1)
df['mRNA_pvalue'] = df.apply(lambda row: calculate_pvalue(row, 'mRNA.Expression'), axis=1)
df['Protein_pvalue'] = df.apply(lambda row: calculate_pvalue(row, 'Protein.Expression'), axis=1)

# Mark statistically significant changes (p < 0.05)
df['CellViability_Significant'] = df['CellViability_pvalue'] < 0.05
df['mRNA_Significant'] = df['mRNA_pvalue'] < 0.05
df['Protein_Significant'] = df['Protein_pvalue'] < 0.05

df
```

```python
# Simple ranking approach
df['Rank_Score'] = ((df['CellViability_log2FC']) + (df['mRNA_log2FC']) + (df['Protein_log2FC']))

# Sort and select top 5
top_genes = df.sort_values('Rank_Score', ascending=False).head(5)

top_genes
```
